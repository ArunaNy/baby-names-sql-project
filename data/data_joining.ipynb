```python
import pandas as pd 

df = pd.read_csv('yob1880.txt', header = None, names = ['first_name', 'sex', 'num'])
df.insert(0, 'year', 1880)

for year in range(1881, 2026):   
    filename = f'yob{year}.txt'
    new_df = pd.read_csv(filename, header = None, names = ['first_name', 'sex', 'num'])
    new_df.insert(0, 'year', year)
    df = pd.concat([df, new_df], ignore_index = True)
print(df)

```
